```python


# ==========================================
# FUNCIÓN PARA CREAR DIETARY PROFILE
# ==========================================
def create_dietary_profile(row):

    profile = []

    if row["is_vegan"] == True:
        profile.append("vegan")

    if row["is_vegetarian"] == True:
        profile.append("vegetarian")

    if row["is_halal"] == True:
        profile.append("halal")

    if row["is_kosher"] == True:
        profile.append("kosher")

    if row["is_nut_free"] == True:
        profile.append("nut_free")

    if row["is_dairy_free"] == True:
        profile.append("dairy_free")

    if row["is_gluten_free"] == True:
        profile.append("gluten_free")

    # ==========================================
    # si no tiene ninguna categoría
    # ==========================================
    if len(profile) == 0:
        profile.append("unspecified")

    return json.dumps(profile)


# ==========================================
# CREAR NUEVA COLUMNA
# ==========================================
df["dietary_profile_updated"] = df.apply(
    create_dietary_profile,
    axis=1
)


# ==========================================
# VER RESULTADOS
# ==========================================
df[[
    "dietary_profile",
    "dietary_profile_updated",
    "is_vegan",
    "is_vegetarian",
    "is_dairy_free"
]].head(10)


```

```python
# vamos a limpiar los ingredientes para transformar y separar por comas en lugar de tener en raw


# ==========================================
# FUNCIÓN PARA LIMPIAR INGREDIENTES
# ==========================================
def clean_ingredient(text):

    text = text.lower()

    # ==========================================
    # quitar contenido entre paréntesis
    # ejemplo:
    # (14 ounce)
    # (6 ounce)
    # ==========================================
    text = re.sub(r'\([^)]*\)', '', text)

    # ==========================================
    # quitar cantidades al inicio
    # ejemplos:
    # 1
    # 1/2
    # 1 1/2
    # 2.5
    # ¼
    # ½
    # ==========================================
    text = re.sub(
        r'^\s*(\d+\s\d+/\d+|\d+/\d+|\d+(\.\d+)?|[¼½¾⅓⅔⅛⅜⅝⅞]+)\s*',
        '',
        text
    )

    # ==========================================
    # quitar unidades comunes
    # ==========================================
    units = [
        "cup", "cups",
        "tablespoon", "tablespoons",
        "teaspoon", "teaspoons",
        "ounce", "ounces",
        "pound", "pounds",
        "lb", "lbs",
        "can", "cans",
        "container", "containers",
        "package", "packages",
        "strip", "strips",
        "pint", "pints",
        "quart", "quarts",
        "gallon", "gallons",
        "jar", "jars",
        "head", "heads",
        "link", "links",
        "pinch", "pinches"
    ]

    pattern = r'\b(' + '|'.join(units) + r')\b'
    text = re.sub(pattern, '', text)

    # ==========================================
    # quitar fracciones unicode restantes
    # ==========================================
    text = re.sub(r'^[¼½¾⅓⅔⅛⅜⅝⅞]+\s*', '', text)

    # ==========================================
    # reemplazar " and " por coma
    # ==========================================
    text = re.sub(r'\sand\s', ', ', text)

    # ==========================================
    # limpiar espacios extras
    # ==========================================
    text = re.sub(r'\s+', ' ', text)

    # ==========================================
    # limpiar comas extras
    # ==========================================
    text = re.sub(r'\s*,\s*', ', ', text)
    text = re.sub(r',+', ',', text)

    text = text.strip(" ,")

    # ==========================================
    # eliminar strings basura
    # ==========================================
    if len(text) < 3:
        return ""

    bad_endings = [" or", " and"]

    for ending in bad_endings:
        if text.endswith(ending):
            text = text[:-len(ending)].strip()

    return text


# ==========================================
# CREAR COLUMNA PROCESADA
# ==========================================
def process_ingredients(ingredients_json):

    ingredients_list = json.loads(ingredients_json)

    cleaned = [clean_ingredient(i) for i in ingredients_list]

    # eliminar vacíos
    cleaned = [i for i in cleaned if i != ""]

    return ", ".join(cleaned)


df["ingredients_text_processed"] = df["ingredients"].apply(process_ingredients)


listas = df["ingredients_text_processed"].str.split(",")

valores_unicos = set()

for lista in listas:

    ingredientes_limpios = [
        i.strip()
        for i in lista
        if i.strip() != ""
    ]

    valores_unicos.update(ingredientes_limpios)

# ==========================================
# eliminar strings vacíos por seguridad
# ==========================================
valores_unicos.discard("")

# ==========================================
# resultados
# ==========================================
print("Total únicos:", len(valores_unicos))

print(valores_unicos)


df["ingredients_text_processed"] = df["ingredients"].apply(process_ingredients)
```

```python
import json

# ==========================================
# FUNCIÓN PARA EXTRAER VALORES ÚNICOS
# DE COLUMNAS JSON TIPO LISTA
# ==========================================
def get_unique_values(column_name):

    valores_unicos = set()

    for item in df[column_name]:

        try:
            lista = json.loads(item)

            for valor in lista:

                valor = str(valor).strip().lower()

                if valor != "":
                    valores_unicos.add(valor)

        except:
            pass

    return sorted(valores_unicos)


# ==========================================
# COLUMNAS JSON/LISTA A ANALIZAR
# ==========================================
json_columns = [
    "ingredients",
    "directions",
    "ingredients_raw",
    "directions_raw",
    "ingredients_canonical",
    "cuisine_list",
    "course_list",
    "tastes",
    "dietary_profile",
    "dietary_profile_updated",
    "health_flags"
]


# ==========================================
# ANALIZAR TODAS LAS COLUMNAS
# ==========================================
for col in json_columns:

    print("\n" + "="*80)
    print(f"COLUMNA: {col}")

    valores = get_unique_values(col)

    print(f"\nTotal valores únicos: {len(valores)}")

    print("\nPrimeros 10 valores:")
    print(valores[:10])
```

```python

# queremos dificultad o no?
# vamos a hacer las dos opciones



# ==========================================
# FUNCIÓN AUXILIAR
# convierte listas JSON a texto
# ==========================================
def json_list_to_text(value):

    try:
        lista = json.loads(value)
        return ", ".join(lista)

    except:
        return str(value)


# ==========================================
# OPCIÓN 1
# CON DIFICULTAD
# ==========================================
def build_embedding_text_with_difficulty(row):

    cuisines = json_list_to_text(row["cuisine_list"])
    courses = json_list_to_text(row["course_list"])
    tastes = json_list_to_text(row["tastes"])
    dietary = json_list_to_text(row["dietary_profile_updated"])
    health_flags = json_list_to_text(row["health_flags"])

    text = f"""
    Title: {row['recipe_title']}

    Category: {row['category']}
    Subcategory: {row['subcategory']}

    Description:
    {row['description']}

    Ingredients:
    {row['ingredients_text_processed']}

    Directions:
    {row['directions_text']}

    Cuisine:
    {cuisines}

    Course:
    {courses}

    Tastes:
    {tastes}

    Dietary profile:
    {dietary}

    Health flags:
    {health_flags}

    Main ingredient:
    {row['main_ingredient']}

    Cook speed:
    {row['cook_speed']}

    Difficulty:
    {row['difficulty']}
    """

    return " ".join(text.split())


# ==========================================
# OPCIÓN 2
# SIN DIFICULTAD
# ==========================================
def build_embedding_text_without_difficulty(row):

    cuisines = json_list_to_text(row["cuisine_list"])
    courses = json_list_to_text(row["course_list"])
    tastes = json_list_to_text(row["tastes"])
    dietary = json_list_to_text(row["dietary_profile_updated"])
    health_flags = json_list_to_text(row["health_flags"])

    text = f"""
    Title: {row['recipe_title']}

    Category: {row['category']}
    Subcategory: {row['subcategory']}

    Description:
    {row['description']}

    Ingredients:
    {row['ingredients_text_processed']}

    Directions:
    {row['directions_text']}

    Cuisine:
    {cuisines}

    Course:
    {courses}

    Tastes:
    {tastes}

    Dietary profile:
    {dietary}

    Health flags:
    {health_flags}

    Main ingredient:
    {row['main_ingredient']}

    Cook speed:
    {row['cook_speed']}
    """

    return " ".join(text.split())


# ==========================================
# CREAR COLUMNAS
# ==========================================
df_clean["embedding_text_with_difficulty"] = df_clean.apply(
    build_embedding_text_with_difficulty,
    axis=1
)

df_clean["embedding_text_without_difficulty"] = df_clean.apply(
    build_embedding_text_without_difficulty,
    axis=1
)


# ==========================================
# VER EJEMPLOS
# ==========================================
print("\n========== WITH DIFFICULTY ==========\n")
print(df_clean["embedding_text_with_difficulty"].iloc[0])

print("\n\n========== WITHOUT DIFFICULTY ==========\n")
print(df_clean["embedding_text_without_difficulty"].iloc[0])
```

Ya hice una limpieza y análisis bastante completo del dataset.

Primero revisé todas las columnas para entender cuáles eran útiles, cuáles eran raw/redundantes y cómo estaba estructurada la info.

Después limpié ingredientes y armé una nueva columna (`ingredients_text_processed`) para dejar los ingredientes más limpios y útiles para NLP/embeddings, quitando cantidades, unidades, ruido, etc.

También encontré que el `dietary_profile` estaba incompleto (no incluía vegetarian ni dairy_free aunque sí existían los booleanos), así que hice una nueva columna `dietary_profile_updated` usando todos los flags reales (`is_vegan`, `is_vegetarian`, etc.) para que quede consistente.

Luego hice un dataframe limpio (`df_clean`) quitando columnas redundantes o que realmente no aportaban mucho.

Y al final armé dos columnas especiales para embeddings:

* `embedding_text_with_difficulty`
* `embedding_text_without_difficulty`

Las dos juntan toda la info importante de la receta en un solo texto:
título, ingredientes, pasos, dieta, tastes, cuisine, cook speed, etc.

La única diferencia es que una incluye dificultad y la otra no, para que podamos probar cuál funciona mejor.

Con eso ya se pueden empezar a hacer embeddings, similarity search o conectar GPT-2/Sentence Transformers, más o menos como el demo/dummy que vimos con Mario donde escribes algo tipo “quiero algo rápido y vegetariano” y el sistema busca recetas parecidas semánticamente.
